## **Motivações e justificativa**

Nas arquiteturas digitais contemporâneas, a identidade consolidou-se como o principal vetor de exposição e, consequentemente, como o **“novo perímetro”** de segurança. Evidências empíricas em relatórios de referência reforçam esse diagnóstico: no [Verizon Data Breach Investigations Report (DBIR) 2024](https://www.verizon.com/business/resources/reports/2024-dbir-data-breach-investigations-report.pdf), o uso de credenciais roubadas aparece como a principal ação inicial em violações, respondendo por **24%** dos casos analisados. Em paralelo, o mesmo relatório aponta que o chamado **“elemento humano”** permanece dominante como componente de incidentes (envolvendo, por exemplo, phishing e vitimização), o que ajuda a explicar por que controles centrados em identidade se tornaram críticos para reduzir superfície de ataque (VERIZON BUSINESS, 2024).

Esse risco não é apenas frequente; ele também tende a ser operacionalmente custoso. O [Cost of a Data Breach Report 2024 (IBM)](https://newsroom.ibm.com/2024-07-30-ibm-report-escalating-data-breach-disruption-pushes-costs-to-new-highs) identifica credenciais comprometidas/roubadas como o vetor inicial mais comum, em **16%** dos incidentes estudados, e observa ainda que esse tipo de violação está associado aos maiores tempos médios de identificação e contenção (quase 10 meses, segundo o comunicado do relatório), o que amplia a janela de oportunidade para persistência do adversário e aumento do impacto.  Em outras palavras, credenciais comprometidas combinam alta incidência com alta fricção operacional, reforçando sua prioridade em estratégias de detecção e resposta (IBM, 2024).

Nesse contexto, o **“momento do login”** torna-se um ponto privilegiado para a detecção de sinais precoces de comprometimento, frequentemente antes que o impacto se materialize em movimentos laterais, exfiltração de dados ou fraude. Do ponto de vista analítico, logs de autenticação se caracterizam naturalmente como **séries temporais** orientadas a eventos. Os registros de *login* apresentam ordenação, frequência, dependência temporal e contexto (por exemplo, presença de MFA, falha ou sucesso, desafios de autenticação e tipos de fluxo), configurando um domínio particularmente adequado a técnicas de modelagem sequencial. Em vez de tratar ocorrências de forma isolada, o projeto enfatiza trajetórias comportamentais, considerando como padrões evoluem ao longo do tempo e como pequenas mudanças de rotina podem indicar risco. Essa abordagem é especialmente relevante porque ataques associados a contas comprometidas raramente são evidenciados por uma única variável; com mais frequência, emergem como mudanças de padrão, tais como acesso em horários atípicos, alternância incomum de localidades, sequências anômalas de eventos (falha → desafio → sucesso) ou velocidades de deslocamento incompatíveis com as limitações físicas (REN; YE; LI, 2017; MALEKI; MALEKI; JENNINGS, 2021).

A motivação operacional também é direta: centros de operações de segurança (**SOC**) convivem com altos volumes de alertas, dos quais apenas uma fração é efetivamente acionável. A consequência é a sobrecarga de triagem e o aumento do tempo até a detecção e contenção, impactando indicadores como **MTTA** e **MTTR**. Um artefato analítico capaz de produzir um *score* de risco acompanhado de justificativa interpretável contribui para reduzir ruído, priorizar investigações e orientar resposta com maior eficiência, traduzindo ganhos concretos na operação de segurança. A relevância de contas e privilégios nesse encadeamento também aparece em análises de postura ofensiva e **“attack paths”**: o [Microsoft Digital Defense Report 2024](https://www.microsoft.com/en-us/security/security-insider/threat-landscape/microsoft-digital-defense-report-2024) reporta que **61%** dos caminhos de ataque analisados levam a uma conta de usuário sensível, evidenciando como a progressão do atacante frequentemente converge para identidade e permissões (MICROSOFT, 2024).

Entre os sinais de maior valor prático, destacam-se padrões reconhecidos e amplamente utilizados pela indústria por sua interpretabilidade e utilidade imediata, mesmo antes do emprego de modelos de aprendizagem profunda. O fenômeno de **Impossible Travel**, deslocamentos geográficos inconsistentes com o intervalo temporal entre autenticações, explosões de falhas em janelas curtas (***bursts***) e cadeias raras de eventos de autenticação constituem indicadores de alto poder explicativo e forte aderência ao trabalho cotidiano de analistas, pois conectam evidência técnica a hipóteses operacionais verificáveis (MICROSOFT, 2025).

A justificativa técnica do projeto decorre, portanto, da superioridade de modelar comportamento em detrimento de eventos pontuais. Métodos de **séries temporais**, janelas deslizantes, medidas de entropia, modelos de transição (como cadeias de Markov) e arquiteturas sequenciais (como **LSTM**) são apropriados porque capturam dependência temporal (relações entre antes e depois), distinguem regularidade de desvio por meio de baselines individualizados e permitem identificar anomalias contextuais, isto é, situações que podem ser normais para um usuário e atípicas para outro. Essa personalização é crucial para reduzir falsos positivos e melhorar a robustez do sistema em ambientes heterogêneos (REN; YE; LI, 2017; MALEKI; MALEKI; JENNINGS, 2021).

Sob a ótica de negócio e risco, detectar precocemente abuso de credenciais reduz a probabilidade e o impacto de fraudes e acessos indevidos, diminui indisponibilidades associadas a resposta a incidentes, reduz custos de investigação e retrabalho e mitiga exposição reputacional e risco regulatório. Adicionalmente, o projeto favorece a criação de um ativo reutilizável: um motor de risco aplicável a diferentes superfícies de autenticação, como VPN, IAM, SSO e aplicações internas, ampliando a utilidade do investimento além do caso de uso inicial (VERIZON BUSINESS, 2024; IBM, 2024).

No plano metodológico, a proposta se sustenta por uma abordagem em camadas, que equilibra explicabilidade e capacidade preditiva, ao mesmo tempo em que viabiliza maturidade incremental. Regras explícitas e interpretáveis oferecem rápida implementação e defensabilidade em auditorias; modelos clássicos tendem a melhorar generalização e reduzir falsos positivos; e modelos sequenciais (**Markov/LSTM**) capturam sutilezas comportamentais que não se manifestam em métricas agregadas. Essa progressão estabelece um caminho claro de evolução tecnológica e de validação científica, alinhando rigor acadêmico com entregas operacionais (REN; YE; LI, 2017; MALEKI; MALEKI; JENNINGS, 2021).

A adequação dos dados reforça a viabilidade do estudo. A combinação de dados pseudonimizados com enriquecimento geográfico permite quantificar deslocamentos com base em latitude, longitude e tempo, incorporar incerteza por meio do *accuracy_radius_km* e explorar o contexto de autenticação (MFA, desafios, falhas), além de utilizar sinais auxiliares, como a marcação **“É suspeito”**, para apoiar processos de validação e avaliação. Em conjunto, esses elementos oferecem base suficiente para construir features robustas, testar hipóteses e mensurar desempenho de forma consistente (MALEKI; MALEKI; JENNINGS, 2021).

A dimensão extensionista do projeto se alinha a objetivos de desenvolvimento sustentável ao fortalecer a resiliência e a segurança de infraestrutura digital baseada em identidade (ODS 9) e ao reduzir desperdício operacional por meio da diminuição de alertas inúteis, horas improdutivas e custos de resposta (ODS 12). A disponibilização pública de código, documentação e um conjunto de dados devidamente pseudonimizado amplia o impacto social e científico do trabalho, favorecendo a replicabilidade, o avanço da pesquisa aplicada e a transferência de conhecimento para a comunidade acadêmica e profissionais de segurança.
